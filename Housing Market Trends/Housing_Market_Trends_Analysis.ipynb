```python
# ============================================================
# HOUSING MARKET TRENDS ANALYSIS
# Data Science Internship Project
# ============================================================

# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

# Visualization style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

FILE_PATH = "dataset/housing.csv"

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f"Dataset not found at: {FILE_PATH}\n"
        "Please place your CSV file inside the dataset folder."
    )

df = pd.read_csv(FILE_PATH)

print("\nDataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# 3. DISPLAY FUNCTION
# ============================================================

# This makes the code work both in Jupyter/Colab and normal Python.

try:
    from IPython.display import display
except ImportError:
    def display(data):
        print(data)


# ============================================================
# 4. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("FIRST 5 RECORDS")
print("=" * 60)

display(df.head())


print("\n" + "=" * 60)
print("DATASET SHAPE")
print("=" * 60)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])


print("\n" + "=" * 60)
print("COLUMN NAMES")
print("=" * 60)

print(df.columns.tolist())


print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

df.info()


print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

display(df.describe(include="all"))


# ============================================================
# 5. STANDARDIZE COLUMN NAMES
# ============================================================

# Remove unnecessary spaces from column names

df.columns = (
    df.columns
    .str.strip()
)

print("\nColumn names cleaned successfully!")


# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print(
    "\nNumber of duplicate records:",
    duplicate_count
)

if duplicate_count > 0:

    df = df.drop_duplicates()

    print(
        "Duplicate records removed."
    )

else:

    print(
        "No duplicate records found."
    )


# ============================================================
# 7. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": (
        missing_values / len(df) * 100
    ).round(2)
})

display(
    missing_table[
        missing_table["Missing Values"] > 0
    ]
)


# ============================================================
# 8. COLUMN DETECTION
# ============================================================

# Create a normalized column dictionary

column_map = {
    col.lower().strip().replace("_", " "): col
    for col in df.columns
}


def find_column(possible_names):
    """
    Finds the first matching column from the dataset.
    """

    for name in possible_names:

        normalized_name = (
            name.lower()
            .strip()
            .replace("_", " ")
        )

        if normalized_name in column_map:

            return column_map[normalized_name]

    return None


# ------------------------------------------------------------
# PRICE COLUMN
# ------------------------------------------------------------

price_col = find_column([
    "price",
    "house price",
    "sale price",
    "selling price",
    "median house value",
    "median_house_value",
    "property price",
    "house_value"
])


# ------------------------------------------------------------
# AREA COLUMN
# ------------------------------------------------------------

area_col = find_column([
    "area",
    "sqft",
    "sq ft",
    "square feet",
    "square footage",
    "sqft living",
    "sqft_living",
    "living area",
    "property area",
    "total sqft",
    "total area"
])


# ------------------------------------------------------------
# BEDROOM COLUMN
# ------------------------------------------------------------

bedrooms_col = find_column([
    "bedrooms",
    "bedroom",
    "beds",
    "number of bedrooms",
    "total bedrooms",
    "total_bedrooms"
])


# ------------------------------------------------------------
# BATHROOM COLUMN
# ------------------------------------------------------------

bathrooms_col = find_column([
    "bathrooms",
    "bathroom",
    "baths",
    "number of bathrooms"
])


# ------------------------------------------------------------
# LOCATION COLUMN
# ------------------------------------------------------------

location_col = find_column([
    "location",
    "city",
    "neighborhood",
    "locality",
    "area name",
    "area_name",
    "district",
    "region",
    "ocean proximity",
    "ocean_proximity"
])


# ------------------------------------------------------------
# PROPERTY TYPE
# ------------------------------------------------------------

property_type_col = find_column([
    "property type",
    "property_type",
    "type",
    "house type",
    "housing type"
])


# ------------------------------------------------------------
# SALE YEAR
# ------------------------------------------------------------

sale_year_col = find_column([
    "sale year",
    "year sold",
    "sold year",
    "transaction year",
    "year"
])


# ------------------------------------------------------------
# BUILD YEAR
# ------------------------------------------------------------

build_year_col = find_column([
    "year built",
    "built year",
    "yr built",
    "yr_built",
    "construction year"
])


# ------------------------------------------------------------
# PARKING
# ------------------------------------------------------------

parking_col = find_column([
    "parking",
    "garage",
    "garage spaces",
    "parking spaces"
])


# ------------------------------------------------------------
# DISPLAY DETECTED COLUMNS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DETECTED COLUMNS")
print("=" * 60)

print("Price          :", price_col)
print("Area           :", area_col)
print("Bedrooms       :", bedrooms_col)
print("Bathrooms      :", bathrooms_col)
print("Location       :", location_col)
print("Property Type  :", property_type_col)
print("Sale Year      :", sale_year_col)
print("Build Year     :", build_year_col)
print("Parking        :", parking_col)


# ============================================================
# 9. VALIDATE PRICE COLUMN
# ============================================================

if price_col is None:

    raise ValueError(
        "\nCould not identify the house price column.\n"
        "Please check your dataset and add its column name "
        "to the price_col list."
    )


# ============================================================
# 10. CONVERT NUMERICAL COLUMNS
# ============================================================

numeric_columns = [
    price_col,
    area_col,
    bedrooms_col,
    bathrooms_col,
    sale_year_col,
    build_year_col,
    parking_col
]

for col in numeric_columns:

    if col is not None:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )


# ============================================================
# 11. REMOVE INVALID PRICE VALUES
# ============================================================

before_cleaning = len(df)

df = df[
    df[price_col].notna()
]

df = df[
    df[price_col] > 0
]

after_cleaning = len(df)

print(
    "\nInvalid price records removed:",
    before_cleaning - after_cleaning
)


# ============================================================
# 12. REMOVE INVALID AREA VALUES
# ============================================================

if area_col is not None:

    df = df[
        df[area_col].isna() |
        (df[area_col] > 0)
    ]


# ============================================================
# 13. FINAL DATASET AFTER CLEANING
# ============================================================

print("\n" + "=" * 60)
print("CLEANED DATASET")
print("=" * 60)

print(
    "Rows after cleaning   :",
    df.shape[0]
)

print(
    "Columns after cleaning:",
    df.shape[1]
)

display(df.head())


# ============================================================
# 14. BASIC PRICE STATISTICS
# ============================================================

average_price = df[price_col].mean()
median_price = df[price_col].median()
highest_price = df[price_col].max()
lowest_price = df[price_col].min()

print("\n" + "=" * 60)
print("HOUSING PRICE STATISTICS")
print("=" * 60)

print(
    f"Average House Price : {average_price:,.2f}"
)

print(
    f"Median House Price  : {median_price:,.2f}"
)

print(
    f"Highest House Price : {highest_price:,.2f}"
)

print(
    f"Lowest House Price  : {lowest_price:,.2f}"
)


# ============================================================
# 15. HOUSE PRICE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    df[price_col],
    bins=30,
    kde=True
)

plt.title(
    "Distribution of House Prices"
)

plt.xlabel(
    "House Price"
)

plt.ylabel(
    "Number of Properties"
)

plt.tight_layout()
plt.show()


# ============================================================
# 16. HOUSE PRICE BOX PLOT
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df[price_col]
)

plt.title(
    "House Price Distribution and Outliers"
)

plt.xlabel(
    "House Price"
)

plt.tight_layout()
plt.show()


# ============================================================
# 17. PRICE PER SQUARE FOOT
# ============================================================

if area_col is not None:

    valid_area = (
        df[area_col].notna() &
        (df[area_col] > 0)
    )

    df.loc[
        valid_area,
        "Price_Per_Sqft"
    ] = (
        df.loc[valid_area, price_col] /
        df.loc[valid_area, area_col]
    )

    print("\n" + "=" * 60)
    print("PRICE PER SQUARE FOOT")
    print("=" * 60)

    print(
        "Average Price/Sq.Ft:",
        f"{df['Price_Per_Sqft'].mean():,.2f}"
    )

    print(
        "Median Price/Sq.Ft :",
        f"{df['Price_Per_Sqft'].median():,.2f}"
    )


# ============================================================
# 18. AREA VS HOUSE PRICE
# ============================================================

if area_col is not None:

    plot_data = df[
        [area_col, price_col]
    ].dropna()

    plt.figure(figsize=(10, 7))

    sns.scatterplot(
        data=plot_data,
        x=area_col,
        y=price_col,
        alpha=0.6
    )

    plt.title(
        "Property Area vs House Price"
    )

    plt.xlabel(
        "Property Area"
    )

    plt.ylabel(
        "House Price"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 19. AREA-PRICE CORRELATION
# ============================================================

if area_col is not None:

    correlation = df[
        [area_col, price_col]
    ].corr().iloc[0, 1]

    print(
        "\nCorrelation between area and price:",
        round(correlation, 3)
    )


# ============================================================
# 20. BEDROOM ANALYSIS
# ============================================================

if bedrooms_col is not None:

    bedroom_analysis = (
        df.groupby(
            bedrooms_col
        )[price_col]
        .agg(
            Average_Price="mean",
            Median_Price="median",
            Property_Count="count"
        )
        .sort_index()
    )

    print("\n" + "=" * 60)
    print("PRICE BY NUMBER OF BEDROOMS")
    print("=" * 60)

    display(bedroom_analysis)


# ============================================================
# 21. AVERAGE PRICE BY BEDROOMS
# ============================================================

if bedrooms_col is not None:

    bedroom_price = (
        df.groupby(
            bedrooms_col
        )[price_col]
        .mean()
        .sort_index()
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=bedroom_price.index.astype(str),
        y=bedroom_price.values
    )

    plt.title(
        "Average House Price by Number of Bedrooms"
    )

    plt.xlabel(
        "Number of Bedrooms"
    )

    plt.ylabel(
        "Average House Price"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 22. BATHROOM ANALYSIS
# ============================================================

if bathrooms_col is not None:

    bathroom_price = (
        df.groupby(
            bathrooms_col
        )[price_col]
        .mean()
        .sort_index()
    )

    print("\n" + "=" * 60)
    print("AVERAGE PRICE BY NUMBER OF BATHROOMS")
    print("=" * 60)

    display(bathroom_price)

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=bathroom_price.index.astype(str),
        y=bathroom_price.values
    )

    plt.title(
        "Average House Price by Number of Bathrooms"
    )

    plt.xlabel(
        "Number of Bathrooms"
    )

    plt.ylabel(
        "Average House Price"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 23. LOCATION ANALYSIS
# ============================================================

if location_col is not None:

    location_analysis = (
        df.groupby(
            location_col
        )[price_col]
        .agg(
            Average_Price="mean",
            Median_Price="median",
            Property_Count="count"
        )
        .sort_values(
            "Average_Price",
            ascending=False
        )
    )

    print("\n" + "=" * 60)
    print("PRICE BY LOCATION")
    print("=" * 60)

    display(
        location_analysis.head(15)
    )


# ============================================================
# 24. MOST EXPENSIVE LOCATIONS
# ============================================================

if location_col is not None:

    top_locations = (
        df.groupby(
            location_col
        )[price_col]
        .mean()
        .sort_values(
            ascending=False
        )
        .head(10)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=top_locations.values,
        y=top_locations.index
    )

    plt.title(
        "Top 10 Most Expensive Locations"
    )

    plt.xlabel(
        "Average House Price"
    )

    plt.ylabel(
        "Location"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 25. MOST AFFORDABLE LOCATIONS
# ============================================================

if location_col is not None:

    affordable_locations = (
        df.groupby(
            location_col
        )[price_col]
        .mean()
        .sort_values()
        .head(10)
    )

    print("\n" + "=" * 60)
    print("MOST AFFORDABLE LOCATIONS")
    print("=" * 60)

    display(affordable_locations)

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=affordable_locations.values,
        y=affordable_locations.index
    )

    plt.title(
        "10 Most Affordable Locations"
    )

    plt.xlabel(
        "Average House Price"
    )

    plt.ylabel(
        "Location"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 26. PROPERTY TYPE ANALYSIS
# ============================================================

if property_type_col is not None:

    property_type_analysis = (
        df.groupby(
            property_type_col
        )[price_col]
        .agg(
            Average_Price="mean",
            Median_Price="median",
            Property_Count="count"
        )
        .sort_values(
            "Average_Price",
            ascending=False
        )
    )

    print("\n" + "=" * 60)
    print("PRICE BY PROPERTY TYPE")
    print("=" * 60)

    display(property_type_analysis)


# ============================================================
# 27. PROPERTY TYPE VISUALIZATION
# ============================================================

if property_type_col is not None:

    property_type_price = (
        df.groupby(
            property_type_col
        )[price_col]
        .mean()
        .sort_values(
            ascending=False
        )
        .head(10)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=property_type_price.values,
        y=property_type_price.index
    )

    plt.title(
        "Average House Price by Property Type"
    )

    plt.xlabel(
        "Average House Price"
    )

    plt.ylabel(
        "Property Type"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 28. YEAR-WISE PRICE ANALYSIS
# ============================================================

# Only use an actual sale/transaction year.
# Do NOT use construction year as market trend year.

if sale_year_col is not None:

    yearly_price = (
        df.groupby(
            sale_year_col
        )[price_col]
        .agg(
            Average_Price="mean",
            Median_Price="median",
            Property_Count="count"
        )
        .sort_index()
    )

    print("\n" + "=" * 60)
    print("YEAR-WISE HOUSING MARKET ANALYSIS")
    print("=" * 60)

    display(
        yearly_price.tail(20)
    )


    # --------------------------------------------------------
    # PRICE TREND
    # --------------------------------------------------------

    plt.figure(figsize=(14, 6))

    plt.plot(
        yearly_price.index,
        yearly_price["Average_Price"],
        marker="o"
    )

    plt.title(
        "Average House Price Trend Over the Years"
    )

    plt.xlabel(
        "Sale Year"
    )

    plt.ylabel(
        "Average House Price"
    )

    plt.tight_layout()
    plt.show()


    # --------------------------------------------------------
    # NUMBER OF PROPERTIES
    # --------------------------------------------------------

    plt.figure(figsize=(14, 6))

    plt.plot(
        yearly_price.index,
        yearly_price["Property_Count"],
        marker="o"
    )

    plt.title(
        "Number of Properties by Sale Year"
    )

    plt.xlabel(
        "Sale Year"
    )

    plt.ylabel(
        "Number of Properties"
    )

    plt.tight_layout()
    plt.show()


else:

    print(
        "\nSale year column was not found."
    )

    print(
        "Year-wise market trend analysis skipped."
    )


# ============================================================
# 29. PROPERTY AGE ANALYSIS
# ============================================================

if build_year_col is not None:

    # Use the latest year available in the dataset
    reference_year = int(
        df[build_year_col].dropna().max()
    )

    df["Property_Age"] = (
        reference_year -
        df[build_year_col]
    )

    # Remove invalid ages
    df.loc[
        df["Property_Age"] < 0,
        "Property_Age"
    ] = np.nan


    if df["Property_Age"].notna().any():

        age_price = (
            df.groupby(
                "Property_Age"
            )[price_col]
            .mean()
            .sort_index()
        )

        print("\n" + "=" * 60)
        print("PROPERTY AGE ANALYSIS")
        print("=" * 60)

        display(
            age_price.head(20)
        )

        plt.figure(figsize=(12, 6))

        plt.plot(
            age_price.index,
            age_price.values,
            marker="o"
        )

        plt.title(
            "Property Age vs Average House Price"
        )

        plt.xlabel(
            "Property Age"
        )

        plt.ylabel(
            "Average House Price"
        )

        plt.tight_layout()
        plt.show()


# ============================================================
# 30. PARKING ANALYSIS
# ============================================================

if parking_col is not None:

    parking_price = (
        df.groupby(
            parking_col
        )[price_col]
        .mean()
        .sort_index()
    )

    print("\n" + "=" * 60)
    print("PRICE BY PARKING AVAILABILITY")
    print("=" * 60)

    display(parking_price)

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=parking_price.index.astype(str),
        y=parking_price.values
    )

    plt.title(
        "Average House Price by Parking Spaces"
    )

    plt.xlabel(
        "Parking Spaces"
    )

    plt.ylabel(
        "Average House Price"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 31. CORRELATION ANALYSIS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns

if len(numeric_columns) >= 2:

    correlation_matrix = df[
        numeric_columns
    ].corr()

    print("\n" + "=" * 60)
    print("CORRELATION MATRIX")
    print("=" * 60)

    display(correlation_matrix)

    plt.figure(figsize=(12, 9))

    sns.heatmap(
        correlation_matrix,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.5
    )

    plt.title(
        "Housing Market Correlation Heatmap"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 32. MOST EXPENSIVE PROPERTIES
# ============================================================

print("\n" + "=" * 60)
print("TOP 10 MOST EXPENSIVE PROPERTIES")
print("=" * 60)

expensive_properties = (
    df.sort_values(
        by=price_col,
        ascending=False
    )
    .head(10)
)

display(expensive_properties)


# ============================================================
# 33. MOST AFFORDABLE PROPERTIES
# ============================================================

print("\n" + "=" * 60)
print("TOP 10 MOST AFFORDABLE PROPERTIES")
print("=" * 60)

affordable_properties = (
    df.sort_values(
        by=price_col,
        ascending=True
    )
    .head(10)
)

display(affordable_properties)


# ============================================================
# 34. PRICE CATEGORIES
# ============================================================

# qcut can fail if there are too few unique values.
# Use duplicates="drop" and determine labels dynamically.

if df[price_col].nunique() >= 4:

    df["Price_Category"] = pd.qcut(
        df[price_col],
        q=4,
        labels=[
            "Budget",
            "Affordable",
            "Premium",
            "Luxury"
        ],
        duplicates="drop"
    )

    print("\n" + "=" * 60)
    print("HOUSING PRICE CATEGORIES")
    print("=" * 60)

    display(
        df["Price_Category"]
        .value_counts()
        .sort_index()
    )

    plt.figure(figsize=(10, 6))

    sns.countplot(
        data=df,
        x="Price_Category",
        order=[
            "Budget",
            "Affordable",
            "Premium",
            "Luxury"
        ]
    )

    plt.title(
        "Distribution of Housing Price Categories"
    )

    plt.xlabel(
        "Price Category"
    )

    plt.ylabel(
        "Number of Properties"
    )

    plt.tight_layout()
    plt.show()

else:

    print(
        "\nNot enough unique prices to create "
        "four price categories."
    )


# ============================================================
# 35. AREA CATEGORIES
# ============================================================

if area_col is not None:

    valid_area_count = (
        df[area_col]
        .dropna()
        .nunique()
    )

    if valid_area_count >= 4:

        df["Area_Category"] = pd.qcut(
            df[area_col],
            q=4,
            labels=[
                "Small",
                "Medium",
                "Large",
                "Very Large"
            ],
            duplicates="drop"
        )

        print("\n" + "=" * 60)
        print("PROPERTY AREA CATEGORIES")
        print("=" * 60)

        display(
            df["Area_Category"]
            .value_counts()
            .sort_index()
        )


# ============================================================
# 36. AREA CATEGORY VS HOUSE PRICE
# ============================================================

if (
    area_col is not None and
    "Area_Category" in df.columns
):

    area_category_price = (
        df.groupby(
            "Area_Category",
            observed=True
        )[price_col]
        .mean()
    )

    print("\n" + "=" * 60)
    print("AVERAGE PRICE BY PROPERTY SIZE")
    print("=" * 60)

    display(area_category_price)

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=area_category_price.index.astype(str),
        y=area_category_price.values
    )

    plt.title(
        "Average House Price by Property Size"
    )

    plt.xlabel(
        "Property Size Category"
    )

    plt.ylabel(
        "Average House Price"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 37. LOCATION + PROPERTY TYPE ANALYSIS
# ============================================================

if (
    location_col is not None and
    property_type_col is not None
):

    location_type_analysis = (
        df.groupby(
            [location_col, property_type_col]
        )[price_col]
        .agg(
            Average_Price="mean",
            Property_Count="count"
        )
        .sort_values(
            "Average_Price",
            ascending=False
        )
        .head(20)
    )

    print("\n" + "=" * 60)
    print("LOCATION + PROPERTY TYPE ANALYSIS")
    print("=" * 60)

    display(location_type_analysis)


# ============================================================
# 38. HIGH-VALUE PROPERTY ANALYSIS
# ============================================================

if area_col is not None:

    median_price = df[price_col].median()
    median_area = df[area_col].median()

    high_value_properties = df[
        (df[price_col] >= median_price) &
        (df[area_col] >= median_area)
    ].copy()

    if len(high_value_properties) > 0:

        high_value_properties["Area_Per_Price"] = (
            high_value_properties[area_col] /
            high_value_properties[price_col]
        )

        high_value_properties = (
            high_value_properties
            .sort_values(
                "Area_Per_Price",
                ascending=False
            )
            .head(10)
        )

        print("\n" + "=" * 60)
        print("POTENTIAL HIGH-VALUE PROPERTIES")
        print("=" * 60)

        display(
            high_value_properties
        )


# ============================================================
# 39. MARKET SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("HOUSING MARKET TRENDS - PROJECT SUMMARY")
print("=" * 65)

print(
    f"Total Properties Analyzed : {len(df):,}"
)

print(
    f"Average Property Price    : "
    f"{df[price_col].mean():,.2f}"
)

print(
    f"Median Property Price     : "
    f"{df[price_col].median():,.2f}"
)

print(
    f"Highest Property Price    : "
    f"{df[price_col].max():,.2f}"
)

print(
    f"Lowest Property Price     : "
    f"{df[price_col].min():,.2f}"
)


if area_col is not None:

    print(
        f"Average Property Area     : "
        f"{df[area_col].mean():,.2f}"
    )


if bedrooms_col is not None:

    print(
        f"Average Bedrooms          : "
        f"{df[bedrooms_col].mean():.2f}"
    )


if bathrooms_col is not None:

    print(
        f"Average Bathrooms         : "
        f"{df[bathrooms_col].mean():.2f}"
    )


if location_col is not None:

    print(
        f"Number of Locations       : "
        f"{df[location_col].nunique()}"
    )


print("=" * 65)


# ============================================================
# 40. AUTOMATIC KEY INSIGHTS
# ============================================================

print("\n" + "=" * 60)
print("KEY INSIGHTS")
print("=" * 60)


# Insight 1
print(
    f"1. The average property price is "
    f"{df[price_col].mean():,.2f}."
)


# Insight 2
print(
    f"2. The median property price is "
    f"{df[price_col].median():,.2f}."
)


# Insight 3
if area_col is not None:

    area_correlation = (
        df[
            [area_col, price_col]
        ]
        .corr()
        .iloc[0, 1]
    )

    print(
        f"3. The correlation between property "
        f"area and price is "
        f"{area_correlation:.3f}."
    )


# Insight 4
if bedrooms_col is not None:

    bedroom_average = (
        df.groupby(
            bedrooms_col
        )[price_col]
        .mean()
    )

    best_bedroom_group = (
        bedroom_average.idxmax()
    )

    print(
        f"4. Properties with "
        f"{best_bedroom_group} bedrooms have "
        f"the highest average price."
    )


# Insight 5
if location_col is not None:

    location_average = (
        df.groupby(
            location_col
        )[price_col]
        .mean()
    )

    most_expensive_location = (
        location_average.idxmax()
    )

    print(
        f"5. {most_expensive_location} has "
        f"the highest average property price."
    )


# Insight 6
if property_type_col is not None:

    property_average = (
        df.groupby(
            property_type_col
        )[price_col]
        .mean()
    )

    most_expensive_type = (
        property_average.idxmax()
    )

    print(
        f"6. {most_expensive_type} has "
        f"the highest average price among "
        f"property types."
    )


# Insight 7
if sale_year_col is not None:

    yearly_average = (
        df.groupby(
            sale_year_col
        )[price_col]
        .mean()
    )

    if len(yearly_average) > 0:

        highest_price_year = (
            yearly_average.idxmax()
        )

        print(
            f"7. The year with the highest "
            f"average property price is "
            f"{highest_price_year}."
        )


# ============================================================
# 41. SAVE CLEANED DATASET
# ============================================================

OUTPUT_FILE = (
    "dataset/cleaned_housing.csv"
)

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    "\nCleaned dataset saved successfully to:"
)

print(
    OUTPUT_FILE
)


# ============================================================
# 42. PROJECT COMPLETION
# ============================================================

print("\n" + "=" * 65)

print(
    "HOUSING MARKET TRENDS ANALYSIS "
    "COMPLETED SUCCESSFULLY!"
)

print("=" * 65)
```
